```python
import random
def seleccion_por_ruleta(poblacion, fitness):
    
    if not poblacion or not fitness or len(poblacion) != len(fitness):
        raise ValueError("Poblacion y fitness deben tener la misma longitud y no estar vacios.")

    # Para evitar fitness negativos, aplicamos un desplazamiento que solvente este problema, respetando la jerarquia 
    # y la diferencia proporcional entre los fitness. Por otra parte, incrementamos 1e-6 para evitar que haya individuos con probabilidad
    # nula de ser seleccionados, favoreciendo asi la diversidad genetica, o que la suma de los fitness sea 0  
    minimo = min(fitness)
    if minimo <= 0:
        fitness = [x - minimo + 1e-6 for x in fitness]
        
    total=sum(fitness)

    punto_aleatorio = random.uniform(0, total)

    aux = 0
    for individuo, individuo_fit in zip(poblacion, fitness):
        aux += individuo_fit
        if aux >= punto_aleatorio:
            return individuo

    # Veamos un ejemplo
def ejemplo():
    poblacion = ["individuo_1", "individuo_2", "individuo_3", "individuo_4"]
    fitness = [3,1,9,7]  # El individuo 3 tiene el mayor fitness, así que es más probable que sea elegido

    # Realizamos 15000 selecciones para comprobar la tendencia
    conteo = {indiv: 0 for indiv in poblacion}
    iteraciones=15000
    for _ in range(iteraciones):
        seleccionado = seleccion_por_ruleta(poblacion, fitness)
        conteo[seleccionado] += 1

    print(f"Resultados tras {iteraciones} iteraciones:")
    for indiv, num_selec in conteo.items():
        porcentaje = (num_selec / iteraciones) * 100
        print(f"El {indiv} ha sido seleccionado {num_selec} veces, es decir, con una frecuencia del {porcentaje:.3f}%")
ejemplo()
```
