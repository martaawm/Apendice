```python
import numpy
class PerceptronSimple:
    def __init__(neuronas, n_entradas, n_neuronas=2, ta=0.1, iteraciones=100): 
        #ta:= tasa de aprendizaje
        #En este caso vamos a considerar una unica capa con dos neuronas.
        #La primera neurona intentara resolver la funcion AND y la segunda la XOR.
        neuronas.w = numpy.random.randn(n_neuronas, n_entradas) #matriz de pesos (cada fila corresponde a una neurona)
        neuronas.b = numpy.random.randn(n_neuronas) #umbrales
        neuronas.ta = ta
        neuronas.iteraciones = iteraciones

    def step(neuronas, z):
        return numpy.where(z >= 0, 1, 0) #definimos la funcion Step

    def propagacion(neuronas, x):
        return numpy.dot(neuronas.w, x) + neuronas.b  #funcion de propagacion
    
    def activacion(neuronas, x):   #función de activacion
        z=neuronas.propagacion(x)      
        return neuronas.step(z)
    
    #tomamos como funcion de salida la identidad

    def entrenar(neuronas, X, y):
        for iteraciones in range(neuronas.iteraciones):
            errores_neurona1 = 0
            errores_neurona2 = 0
            
            for xi, esperado in zip(X, y):
                salida = neuronas.activacion(xi)
                error = esperado - salida 
                
                if error[0] != 0:
                    errores_neurona1 += 1
                if error[1] != 0:
                    errores_neurona2 += 1
                
                # Si alguna de las neuronas comete un error, se actualizan los pesos y el umbral
                if numpy.any(error != 0):
                    neuronas.w += neuronas.ta * numpy.outer(error, xi)  #tasa de aprendizaje*error*entrada
                    neuronas.b += neuronas.ta * error
            
            if errores_neurona1 == 0 and errores_neurona2 != 0:
                print(f"La neurona 1 (AND) ya ha convergido en la iteracion {iteraciones+1}, pero la neurona 2 (XOR) sigue fallando")
            
            if errores_neurona1 == 0 and errores_neurona2 == 0:
                print(f"¡Ambas neuronas convergen en la iteracion {iteraciones+1}!")
                return True
                
        print(f"La neurona 1 ha conseguido resolver AND, mientras que la neurona 2 (XOR) no converge.")
        return False
    
X = numpy.array([[0,0], [0,1], [1,0], [1,1]])
y = numpy.array([
    [0, 0],  # [0,0] -> AND=0, XOR=0
    [0, 1],  # [0,1] -> AND=0, XOR=1
    [0, 1],  # [1,0] -> AND=0, XOR=1
    [1, 0]   # [1,1] -> AND=1, XOR=0
])

capa = PerceptronSimple(n_entradas=2, n_neuronas=2, iteraciones=20)
capa.entrenar(X, y)
```
