```python
import random
import numpy as np
from sklearn.datasets import load_breast_cancer
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.model_selection import GridSearchCV, StratifiedKFold, train_test_split
from sklearn.preprocessing import StandardScaler
from NEAT3 import NEAT
from Retropropagacion import Retropropagacion
from AGsimple import AGsimple

def preprocesar_datos():
    data = load_breast_cancer()
    x = data.data
    y = data.target
    y = 1 - y
    print(f"Los datos recogen {len(x)} observaciones con {x.shape[1]} caracteristicas cada una")
    print(f"Clases: {sum(y==0)} benignos, {sum(y==1)} malignos")
    return x, y

def evaluar_modelo(modelo, x_test, y_test, nombre):
    y_pred = modelo.predict(x_test)
    print(f"Evaluación del aprendizaje del mejor modelo obtenido mediante {nombre} sobre la muestra test")
    print(f"Exactitud: {accuracy_score(y_test, y_pred):.4f}")
    print(f"Precisión: {precision_score(y_test, y_pred, zero_division=0):.4f}")
    print(f"Sensibilidad: {recall_score(y_test, y_pred, zero_division=0):.4f}")
    print(f"F1-score: {f1_score(y_test, y_pred, zero_division=0):.4f}")
    return {"exactitud": accuracy_score(y_test, y_pred), "precision": precision_score(y_test, y_pred, zero_division=0),
            "sensibilidad": recall_score(y_test, y_pred, zero_division=0), "f1_score": f1_score(y_test, y_pred, zero_division=0)}
    
def busqueda(nombre, estimador, param_grid, x_train, y_train, x_test, y_test, cv):
    print(f"\nGridSearchCV para {nombre}")
    gs = GridSearchCV(estimator=estimador, param_grid=param_grid, cv=cv, scoring="accuracy", n_jobs=1, verbose=1, refit=True)
    gs.fit(x_train, y_train)
    print(f"La mejor combinación de hiperparámetros es {gs.best_params_}.")
    print(f"La exactitud media que ha registrado el modelo con dicha combinación durante el entrenamiento es {gs.best_score_:.4f}.")
    medidas = evaluar_modelo(gs, x_test, y_test, nombre)
    return gs, medidas
    
def main():
    x, y = preprocesar_datos()
    x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, stratify=y, random_state=19)
    print(f"\n La muestra de entrenamiento está compuesta por {len(x_train)} observaciones, mientras que la test está compuesta por {len(x_test)} observaciones.")
    scaler = StandardScaler()
    x_train_tipificado = scaler.fit_transform(x_train)
    x_test_tipificado = scaler.transform(x_test)
    cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=19)
    resultados = {}

    param_grid_retro = {"epocas": [10, 30], "lr": [0.001, 0.01], "batches": [1, 32]}
    _, resultados["Retropropagacion"] = busqueda("Retropropagacion", Retropropagacion(), param_grid_retro, x_train_tipificado, y_train, x_test_tipificado, y_test, cv)

    param_grid_AGsimple = {"N": [15, 30], "generaciones": [10, 30], "prob_mut": [0.1,0.2], "prob_cruc": [0.5, 0.75, 0.9]}
    _, resultados["AGsimple"] = busqueda("AGsimple", AGsimple(), param_grid_AGsimple, x_train_tipificado, y_train, x_test_tipificado, y_test, cv)

    param_grid_NEAT = {"N": [15, 30], "generaciones": [10, 30], "delta_t": [1.0, 3.0], "prob_mutar_peso": [0.65, 0.95], "prob_mutar_conexion": [0.03, 0.09], "prob_mutar_nodo": [0.01, 0.05]}
    _, resultados["NEAT"] = busqueda("NEAT", NEAT(), param_grid_NEAT, x_train_tipificado, y_train, x_test_tipificado, y_test, cv)
    
    print("Comparativa sobre la muestra test")
    print(f"{'Métrica':<15} | {'Retroprop':<10} | {'AGsimple':<10} | {'NEAT':<10}")
    print("-" * 55)
    for j in ["exactitud", "precision", "sensibilidad", "f1_score"]:
        fila = f"{j.capitalize():<15} | "
        for nombre in ["Retropropagacion", "AGsimple", "NEAT"]:
            fila += f"{resultados[nombre][j]:<10.4f} | "
        print(fila)
        
if __name__ == "__main__":
    main()
```
