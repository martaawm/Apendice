```python
import math
import random
import numpy as np
import gymnasium as gym
import matplotlib.pyplot as plt
from NEAT1 import Genotipo, Poblacion

# Primero, verifiquemos cuántas observaciones tiene realmente el entorno
def verificar_entorno():
    print("Verificando entornos disponibles...")
    
    # Verificar InvertedDoublePendulum-v4
    try:
        env_v4 = gym.make("InvertedDoublePendulum-v4")
        obs_v4, _ = env_v4.reset()
        print(f"InvertedDoublePendulum-v4:")
        print(f"  Observaciones: {len(obs_v4)}")
        print(f"  Espacio de observación: {env_v4.observation_space.shape}")
        print(f"  Espacio de acción: {env_v4.action_space}")
        env_v4.close()
    except Exception as e:
        print(f"  Error con v4: {e}")
    
    # Verificar InvertedDoublePendulum-v5
    try:
        env_v5 = gym.make("InvertedDoublePendulum-v5")
        obs_v5, _ = env_v5.reset()
        print(f"\nInvertedDoublePendulum-v5:")
        print(f"  Observaciones: {len(obs_v5)}")
        print(f"  Espacio de observación: {env_v5.observation_space.shape}")
        print(f"  Espacio de acción: {env_v5.action_space}")
        env_v5.close()
    except Exception as e:
        print(f"  Error con v5: {e}")
    
    # Verificar Acrobot como alternativa
    try:
        env_acrobot = gym.make("Acrobot-v1")
        obs_acrobot, _ = env_acrobot.reset()
        print(f"\nAcrobot-v1:")
        print(f"  Observaciones: {len(obs_acrobot)}")
        print(f"  Espacio de observación: {env_acrobot.observation_space.shape}")
        print(f"  Espacio de acción: {env_acrobot.action_space}")
        env_acrobot.close()
    except Exception as e:
        print(f"  Error con Acrobot: {e}")

class DoblePenduloGenotipo(Genotipo):
    
    def __init__(self, num_entradas=None, num_salidas=1):
        super().__init__()
        self.num_entradas = num_entradas
        self.num_salidas = num_salidas
        self.fitness = 0.0
    
    def inicializar(self, num_entradas, num_salidas=1):
        self.num_entradas = num_entradas
        self.num_salidas = num_salidas
        return self
    
    def decision(self, observacion):
        # Convertir observación a lista si es necesario
        if isinstance(observacion, np.ndarray):
            observacion = observacion.tolist()
        elif not isinstance(observacion, list):
            observacion = list(observacion)
        
        salida = self.fenotipo(observacion)
        # Para v4, el rango de acción es [-1, 1]
        accion_continua = (salida[0] - 0.5) * 2.0  # Mapeo de [0,1] a [-1,1]
        return np.array([accion_continua], dtype=np.float32)
    
    def evaluar(self, entorno, max_pasos=1000, render=False):
        observacion, _ = entorno.reset()
        pasos = 0
        recompensa_total = 0
        
        for _ in range(max_pasos):
            accion = self.decision(observacion)
            observacion, recompensa, fracaso, maximo, _ = entorno.step(accion)
            pasos += 1
            recompensa_total += recompensa
            
            if fracaso or maximo:
                break
        
        # Fitness normalizado
        self.fitness = (pasos / max_pasos) * 500 + max(0, recompensa_total) * 0.1
        return self.fitness
    
    def calcular_fitness(self, n_evaluaciones=2):
        entorno = gym.make("InvertedDoublePendulum-v4")  # Usamos v4 que es más estable
        fitness_total = 0
        
        for _ in range(n_evaluaciones):
            fitness_total += self.evaluar(entorno, max_pasos=1000)
        
        entorno.close()
        self.fitness = fitness_total / n_evaluaciones
        return self.fitness

class AcrobotGenotipo(Genotipo):
    
    def __init__(self, num_entradas=None, num_salidas=3):
        super().__init__()
        self.num_entradas = num_entradas
        self.num_salidas = num_salidas
        self.fitness = 0.0
    
    def inicializar(self, num_entradas, num_salidas=3):
        self.num_entradas = num_entradas
        self.num_salidas = num_salidas
        return self
    
    def decision(self, observacion):
        if isinstance(observacion, np.ndarray):
            observacion = observacion.tolist()
        
        salidas = self.fenotipo(observacion)
        # Para Acrobot, elegimos la acción con mayor valor
        return int(np.argmax(salidas))
    
    def evaluar(self, entorno, max_pasos=500, render=False):
        observacion, _ = entorno.reset()
        pasos = 0
        recompensa_total = 0
        
        for _ in range(max_pasos):
            accion = self.decision(observacion)
            observacion, recompensa, fracaso, maximo, _ = entorno.step(accion)
            pasos += 1
            recompensa_total += recompensa
            
            if fracaso or maximo:
                break
        
        # Para Acrobot, el objetivo es alcanzar cierta altura
        # Recompensa es -1 por paso hasta alcanzar el objetivo
        self.fitness = -recompensa_total  # Menos pasos = mejor fitness
        return self.fitness
    
    def calcular_fitness(self, n_evaluaciones=3):
        entorno = gym.make("Acrobot-v1")
        fitness_total = 0
        
        for _ in range(n_evaluaciones):
            fitness_total += self.evaluar(entorno, max_pasos=500)
        
        entorno.close()
        self.fitness = fitness_total / n_evaluaciones
        return self.fitness

def entrenar_doble_pendulo(num_generaciones=100, tam=150):
    print("\nEntrenamiento del Doble Péndulo Invertido mediante NEAT\n")
    
    # El entorno v4 tiene 11 observaciones
    poblacion = Poblacion(
        N=tam, 
        genotipos=lambda: DoblePenduloGenotipo(num_entradas=11, num_salidas=1), 
        num_entrada=11, 
        num_salida=1
    )
    
    mejor_fitnesshis = []
    fitness_medhis = []
    verdadero_campeon = None
    
    for generacion in range(num_generaciones):
        fitness_todos = []
        
        for i in poblacion.individuos:
            fitness = i.calcular_fitness(n_evaluaciones=2)
            fitness_todos.append(fitness)
        
        mejor = max(fitness_todos)
        fitness_med = sum(fitness_todos) / len(fitness_todos)
        mejor_fitnesshis.append(mejor)
        fitness_medhis.append(fitness_med)
        
        mejor_ind_gen = max(poblacion.individuos, key=lambda x: x.fitness)
        if verdadero_campeon is None or mejor_ind_gen.fitness > verdadero_campeon.fitness:
            import copy
            verdadero_campeon = copy.deepcopy(mejor_ind_gen)
        
        poblacion.especiacion(delta_t=3.0)
        
        if generacion % 10 == 0 or generacion == num_generaciones - 1:
            nodos_ocultos = sum(1 for n in verdadero_campeon.nodos.values() if n.tipo == 'oculta')
            print(f"Gen {generacion}: fitness medio = {fitness_med:.1f}, mejor = {mejor:.1f}, nodos ocultos = {nodos_ocultos}")
        
        if mejor >= 450:
            print(f"¡Problema resuelto en la generación {generacion}!")
            break
        
        if generacion < num_generaciones - 1:
            poblacion.reproduccion()
    
    return verdadero_campeon, mejor_fitnesshis, fitness_medhis

def entrenar_acrobot(num_generaciones=100, tam=150):
    print("\nEntrenamiento del Acrobot mediante NEAT\n")
    
    # Acrobot tiene 6 observaciones y 3 acciones
    poblacion = Poblacion(
        N=tam, 
        genotipos=lambda: AcrobotGenotipo(num_entradas=6, num_salidas=3), 
        num_entrada=6, 
        num_salida=3
    )
    
    mejor_fitnesshis = []
    fitness_medhis = []
    verdadero_campeon = None
    
    for generacion in range(num_generaciones):
        fitness_todos = []
        
        for i in poblacion.individuos:
            fitness = i.calcular_fitness(n_evaluaciones=3)
            fitness_todos.append(fitness)
        
        mejor = max(fitness_todos)
        fitness_med = sum(fitness_todos) / len(fitness_todos)
        mejor_fitnesshis.append(mejor)
        fitness_medhis.append(fitness_med)
        
        mejor_ind_gen = max(poblacion.individuos, key=lambda x: x.fitness)
        if verdadero_campeon is None or mejor_ind_gen.fitness > verdadero_campeon.fitness:
            import copy
            verdadero_campeon = copy.deepcopy(mejor_ind_gen)
        
        poblacion.especiacion(delta_t=3.0)
        
        if generacion % 10 == 0 or generacion == num_generaciones - 1:
            nodos_ocultos = sum(1 for n in verdadero_campeon.nodos.values() if n.tipo == 'oculta')
            print(f"Gen {generacion}: fitness medio = {fitness_med:.1f}, mejor = {mejor:.1f}, nodos ocultos = {nodos_ocultos}")
        
        # Para Acrobot, mejor fitness = menos pasos para resolver
        if mejor <= 100:  # Resuelto en menos de 100 pasos
            print(f"¡Problema resuelto en la generación {generacion}!")
            break
        
        if generacion < num_generaciones - 1:
            poblacion.reproduccion()
    
    return verdadero_campeon, mejor_fitnesshis, fitness_medhis

def visualizar_resultado(mejor_individuo, tipo_entorno="doble_pendulo", num_intentos=3):
    if tipo_entorno == "doble_pendulo":
        entorno = gym.make("InvertedDoublePendulum-v4", render_mode="human")
    else:  # acrobot
        entorno = gym.make("Acrobot-v1", render_mode="human")
    
    for intento in range(num_intentos):
        print(f"\nIntento {intento + 1}/{num_intentos}")
        observacion, _ = entorno.reset()
        pasos = 0
        recompensa_total = 0
        
        for _ in range(1000):
            accion = mejor_individuo.decision(observacion)
            observacion, recompensa, fracaso, maximo, _ = entorno.step(accion)
            pasos += 1
            recompensa_total += recompensa
            
            if fracaso or maximo:
                break
        
        print(f"Pasos: {pasos}, Recompensa total: {recompensa_total:.1f}")
    
    entorno.close()

def analizar_estructura(individuo):
    nodos_entrada = sum(1 for n in individuo.nodos.values() if n.tipo == 'entrada')
    nodos_ocultos = sum(1 for n in individuo.nodos.values() if n.tipo == 'oculta')
    nodos_salida = sum(1 for n in individuo.nodos.values() if n.tipo == 'salida')
    
    conexiones_activas = sum(1 for c in individuo.conexiones.values() if c.bit_act)
    
    print("\n" + "="*50)
    print("ANÁLISIS DE ESTRUCTURA DE RED")
    print("="*50)
    print(f"Nodos de entrada: {nodos_entrada}")
    print(f"Nodos ocultos: {nodos_ocultos}")
    print(f"Nodos de salida: {nodos_salida}")
    print(f"Conexiones activas: {conexiones_activas}")

def main():
    # Primero verificar entornos
    verificar_entorno()
    
    print("\n" + "="*60)
    print("ELIGE EL PROBLEMA A RESOLVER:")
    print("1. Doble Péndulo Invertido")
    print("2. Acrobot (Péndulo doble con acciones discretas)")
    
    opcion = input("Ingresa 1 o 2: ")
    
    if opcion == "1":
        print("="*60)
        print("ENTRENAMIENTO DE DOBLE PÉNDULO INVERTIDO")
        print("="*60)
        
        mejor_individuo, mejor_fithis, fitness_medhis = entrenar_doble_pendulo(
            num_generaciones=100, 
            tam=150
        )
        visualizar_resultado(mejor_individuo, "doble_pendulo")
    else:
        print("="*60)
        print("ENTRENAMIENTO DE ACROBOT")
        print("="*60)
        
        mejor_individuo, mejor_fithis, fitness_medhis = entrenar_acrobot(
            num_generaciones=100, 
            tam=150
        )
        visualizar_resultado(mejor_individuo, "acrobot")
    
    analizar_estructura(mejor_individuo)
    
    # Graficar evolución
    plt.figure(figsize=(10, 6))
    plt.plot(mejor_fithis, 'b-', label='Mejor fitness')
    plt.plot(fitness_medhis, 'r-', label='Fitness medio')
    plt.xlabel('Generación')
    plt.ylabel('Fitness')
    plt.title('Evolución del Fitness')
    plt.legend()
    plt.grid(True)
    plt.show()

if __name__ == "__main__":
    main()
```
