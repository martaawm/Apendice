```python
import random
import numpy as np
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split, GridSearchCV, StratifiedKFold
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from neat_wrapper import NEATClassifier
from pytorch_wrapper import PyTorchClassifier
from pygad_wrapper import PyGADClassifier


def cargar_datos():
    data = load_breast_cancer()
    X = data.data
    y = 1 - data.target   # maligno -> 1, benigno -> 0
    print(f"Datos: {len(X)} observaciones, {X.shape[1]} características")
    print(f"Clases: {sum(y == 0)} benignos, {sum(y == 1)} malignos")
    return X, y


def evaluar_en_test(modelo, X_test, y_test, nombre):
    y_pred = modelo.predict(X_test)
    print(f"\n--- {nombre} en test ---")
    print(f"  Exactitud:    {accuracy_score(y_test, y_pred):.4f}")
    print(f"  Precisión:    {precision_score(y_test, y_pred, zero_division=0):.4f}")
    print(f"  Sensibilidad: {recall_score(y_test, y_pred, zero_division=0):.4f}")
    print(f"  F1-score:     {f1_score(y_test, y_pred, zero_division=0):.4f}")

    return {
        'exactitud': accuracy_score(y_test, y_pred),
        'precision': precision_score(y_test, y_pred, zero_division=0),
        'sensibilidad': recall_score(y_test, y_pred, zero_division=0),
        'f1_score': f1_score(y_test, y_pred, zero_division=0),
    }


def busqueda_y_evaluacion(nombre, estimador, param_grid, X_train, y_train,
                          X_test, y_test, cv_interna):
    print(f"\n===== GridSearchCV para {nombre} =====")
    gs = GridSearchCV(
        estimator,
        param_grid,
        cv=cv_interna,
        scoring='accuracy',
        n_jobs=1,       # NEAT y PyGAD usan random global, no paralelizar
        verbose=1,
        refit=True,
    )
    gs.fit(X_train, y_train)
    print(f"Mejores hiperparámetros: {gs.best_params_}")
    print(f"Mejor accuracy en CV interna: {gs.best_score_:.4f}")
    medidas = evaluar_en_test(gs, X_test, y_test, nombre)
    return gs, medidas


def main():
    X, y = cargar_datos()

    # División 80/20 estratificada
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.2, stratify=y, random_state=42
    )
    print(f"\nTrain: {len(X_train)} | Test: {len(X_test)}")

    cv_interna = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

    resultados = {}

    # ------------- NEAT -------------
    param_grid_neat = {
        'N': [50, 100],
        'generaciones': [15, 25],
        'delta_t': [2.0, 3.0],
    }
    _, resultados['NEAT'] = busqueda_y_evaluacion(
        "NEAT",
        NEATClassifier(semilla=42, escalar=True),
        param_grid_neat,
        X_train, y_train, X_test, y_test, cv_interna,
    )

    # ------------- PyTorch -------------
    param_grid_torch = {
        'hidden_dim': [8, 16, 32],
        'lr': [0.001, 0.01],
        'epochs': [50, 100],
    }
    _, resultados['PyTorch'] = busqueda_y_evaluacion(
        "PyTorch",
        PyTorchClassifier(semilla=42, escalar=True),
        param_grid_torch,
        X_train, y_train, X_test, y_test, cv_interna,
    )

    # ------------- PyGAD -------------
    param_grid_pygad = {
        'hidden_dim': [4, 8],
        'num_generations': [15, 25],
        'sol_per_pop': [30, 50],
    }
    _, resultados['PyGAD'] = busqueda_y_evaluacion(
        "PyGAD",
        PyGADClassifier(random_seed=42, escalar=True),
        param_grid_pygad,
        X_train, y_train, X_test, y_test, cv_interna,
    )

    # ------------- Resumen -------------
    print("\n==========================================================")
    print("   COMPARATIVA FINAL SOBRE EL CONJUNTO DE TEST (20%)")
    print("==========================================================")
    print(f"{'Métrica':<15} | {'NEAT':<10} | {'PyTorch':<10} | {'PyGAD':<10}")
    print("-" * 55)
    for m in ['exactitud', 'precision', 'sensibilidad', 'f1_score']:
        fila = f"{m.capitalize():<15} | "
        for nombre in ['NEAT', 'PyTorch', 'PyGAD']:
            fila += f"{resultados[nombre][m]:<10.4f} | "
        print(fila)
    print("==========================================================")


if __name__ == "__main__":
    main()
```
